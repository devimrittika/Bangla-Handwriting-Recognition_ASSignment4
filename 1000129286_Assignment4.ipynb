```python
import os
import json
import numpy as np
import tensorflow as tf
from PIL import Image
from sklearn.model_selection import train_test_split

# Dataset paths
images_path = "/content/banglalekha/dataset/BanglaLekha-Isolated/Images"
labels_path = "/content/labels.json"

# Load labels
with open(labels_path, "r", encoding="utf-8") as f:
    labels = json.load(f)

# Image settings
IMG_SIZE = 64

X = []
y = []

print("Loading dataset images...")

for folder in range(1, 85):
    folder_path = os.path.join(images_path, str(folder))
    label_index = folder - 1

    for filename in os.listdir(folder_path):
        if filename.lower().endswith((".png", ".jpg", ".jpeg")):
            image_path = os.path.join(folder_path, filename)

            img = Image.open(image_path).convert("L")
            img = img.resize((IMG_SIZE, IMG_SIZE))

            image_array = np.array(img, dtype=np.float32) / 255.0

            X.append(image_array)
            y.append(label_index)

    if folder % 10 == 0:
        print(f"Processed folder {folder}/84")

# Convert to NumPy arrays
X = np.array(X, dtype=np.float32)
y = np.array(y, dtype=np.int32)

# Add channel dimension
X = X[..., np.newaxis]

print("\nDataset loaded successfully!")
print("Images:", X.shape)
print("Labels:", y.shape)
print("Number of classes:", len(np.unique(y)))
```

```python
import mlflow
import tensorflow as tf
from sklearn.model_selection import train_test_split
from tensorflow.keras import layers, models

# Split dataset: 80% training, 20% validation
X_train, X_val, y_train, y_val = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training images:", X_train.shape)
print("Validation images:", X_val.shape)

# Build CNN model
def build_model():
    model = models.Sequential([
        layers.Input(shape=(64, 64, 1)),

        layers.Conv2D(32, (3, 3), activation="relu"),
        layers.MaxPooling2D((2, 2)),

        layers.Conv2D(64, (3, 3), activation="relu"),
        layers.MaxPooling2D((2, 2)),

        layers.Conv2D(128, (3, 3), activation="relu"),
        layers.MaxPooling2D((2, 2)),

        layers.Flatten(),
        layers.Dense(256, activation="relu"),
        layers.Dropout(0.4),

        layers.Dense(84, activation="softmax")
    ])

    return model


# MLflow experiment
mlflow.set_experiment("Bangla_Handwriting_Recognition")

best_accuracy = 0
best_model = None

# Two experiments with different learning rates
for run_name, learning_rate in [
    ("Experiment_1", 0.001),
    ("Experiment_2", 0.0005)
]:

    print("\nStarting:", run_name)

    with mlflow.start_run(run_name=run_name):

        model = build_model()

        model.compile(
            optimizer=tf.keras.optimizers.Adam(
                learning_rate=learning_rate
            ),
            loss="sparse_categorical_crossentropy",
            metrics=["accuracy"]
        )

        mlflow.log_param("learning_rate", learning_rate)
        mlflow.log_param("batch_size", 128)
        mlflow.log_param("epochs", 3)
        mlflow.log_param("image_size", "64x64")
        mlflow.log_param("model", "CNN")

        history = model.fit(
            X_train,
            y_train,
            validation_data=(X_val, y_val),
            epochs=3,
            batch_size=128,
            verbose=1
        )

        val_accuracy = max(history.history["val_accuracy"])
        val_loss = min(history.history["val_loss"])

        mlflow.log_metric("best_validation_accuracy", float(val_accuracy))
        mlflow.log_metric("best_validation_loss", float(val_loss))

        print("Validation accuracy:", val_accuracy)

        if val_accuracy > best_accuracy:
            best_accuracy = val_accuracy
            best_model = model

# Save best model
best_model.save("/content/model.keras")

print("\nTraining completed!")
print("Best validation accuracy:", best_accuracy)
print("Model saved at /content/model.keras")
```

```python
import tensorflow as tf

images_path = "/content/banglalekha/dataset/BanglaLekha-Isolated/Images"

BATCH_SIZE = 64
IMG_SIZE = (64, 64)

class_names = [str(i) for i in range(1, 85)]

train_ds = tf.keras.utils.image_dataset_from_directory(
    images_path,
    labels="inferred",
    label_mode="int",
    class_names=class_names,
    color_mode="grayscale",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    validation_split=0.2,
    subset="training",
    seed=42
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    images_path,
    labels="inferred",
    label_mode="int",
    class_names=class_names,
    color_mode="grayscale",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    validation_split=0.2,
    subset="validation",
    seed=42
)

# Normalize pixel values
normalization_layer = tf.keras.layers.Rescaling(1./255)

train_ds = train_ds.map(
    lambda x, y: (normalization_layer(x), y)
)

val_ds = val_ds.map(
    lambda x, y: (normalization_layer(x), y)
)

# Improve loading efficiency
AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)

print("Training and validation datasets created!")
print("Number of classes:", len(class_names))
```

```python
import tensorflow as tf
import mlflow

from tensorflow.keras import layers, models

def build_model():
    model = models.Sequential([
        layers.Input(shape=(64, 64, 1)),

        layers.Conv2D(32, 3, activation="relu"),
        layers.MaxPooling2D(),

        layers.Conv2D(64, 3, activation="relu"),
        layers.MaxPooling2D(),

        layers.Conv2D(128, 3, activation="relu"),
        layers.MaxPooling2D(),

        layers.Flatten(),
        layers.Dense(256, activation="relu"),
        layers.Dropout(0.4),

        layers.Dense(84, activation="softmax")
    ])

    return model


mlflow.set_experiment("Bangla_Handwriting_Recognition")

best_accuracy = 0

for run_name, learning_rate in [
    ("Experiment_1", 0.001),
    ("Experiment_2", 0.0005)
]:

    print("\nStarting:", run_name)

    with mlflow.start_run(run_name=run_name):

        model = build_model()

        model.compile(
            optimizer=tf.keras.optimizers.Adam(
                learning_rate=learning_rate
            ),
            loss="sparse_categorical_crossentropy",
            metrics=["accuracy"]
        )

        mlflow.log_param("learning_rate", learning_rate)
        mlflow.log_param("batch_size", 64)
        mlflow.log_param("epochs", 3)
        mlflow.log_param("model", "CNN")

        history = model.fit(
            train_ds,
            validation_data=val_ds,
            epochs=3,
            verbose=1
        )

        val_accuracy = max(history.history["val_accuracy"])

        mlflow.log_metric(
            "best_validation_accuracy",
            float(val_accuracy)
        )

        print("Validation accuracy:", val_accuracy)

        if val_accuracy > best_accuracy:
            best_accuracy = val_accuracy
            model.save("/content/model.keras")

print("\nBoth experiments completed!")
print("Best validation accuracy:", best_accuracy)
print("Best model saved at /content/model.keras")
```
